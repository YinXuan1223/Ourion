## 1. GIF for Camera View

```python
from PIL import Image
import os

# ============ Parameter Setting ============
base_path = "/mnt/HDD1/iccv/orion_workspace/Orion/data/bench2drive/v1"

scene_name = "ConstructionObstacle_Town05_Route68_Weather8"
camera_view = "rgb_front"
frame_start = 6
frame_end = 51
frame_step = 1

frame_duration_ms = 100
output_filename = f"6_10Hz.gif"
# =================================

image_paths = []
for frame_num in range(frame_start, frame_end + 1, frame_step):
    frame_str = f"{frame_num:05d}"
    image_path = os.path.join(base_path, scene_name, "camera", camera_view, f"{frame_str}.jpg")
    image_paths.append(image_path)

frames = []
for path in image_paths:
    if os.path.exists(path):
        frames.append(Image.open(path).convert("RGB"))
        # print(f"Added {path}")
    else:
        print(f"File not found: {path}")

if frames:
    frames[0].save(
        output_filename,
        save_all=True,
        append_images=frames[1:],
        duration=frame_duration_ms,
        loop=0
    )
    print(f"GIF saved to {os.path.abspath(output_filename)}")
```

## 1.1 GIF for Camera view of 1 clip

```python
from PIL import Image
import os

# ============ Parameter Setting ============

scene_name = "ConstructionObstacle_Town05_Route68_Weather8"
frame_start = 28

# =================================

base_path = "/mnt/HDD1/iccv/orion_workspace/Orion/data/bench2drive/v1"
frame_end = frame_start + 45
frame_step = 5
camera_view = "rgb_front"
frame_duration_ms = 500
output_filename = f"{frame_start}.gif"
output_path = f"/mnt/HDD1/iccv/orion_workspace/Bench2Drive/Orion/data/GIFs/clips/front_view/{scene_name}"

image_paths = []
for frame_num in range(frame_start, frame_end + 1, frame_step):
    frame_str = f"{frame_num:05d}"
    image_path = os.path.join(base_path, scene_name, "camera", camera_view, f"{frame_str}.jpg")
    image_paths.append(image_path)

frames = []
for path in image_paths:
    if os.path.exists(path):
        frames.append(Image.open(path).convert("RGB"))
        # print(f"Added {path}")
    else:
        print(f"File not found: {path}")

if frames:
    os.makedirs(output_path, exist_ok=True)
    save_path = os.path.join(output_path, output_filename)
    
    frames[0].save(
        save_path,
        save_all=True,
        append_images=frames[1:],
        duration=frame_duration_ms,
        loop=0
    )
    print(f"GIF saved to {os.path.abspath(save_path)}")
```

## 2. GIF for BEV

```python
from PIL import Image
import os
import glob

# ============ Parameter Setting ============

scene_name = "ConstructionObstacle_Town05_Route68_Weather8"
clip_num = 1

# =========================================

input_dir = f"/mnt/HDD1/iccv/orion_workspace/Orion/data/styleclips/BEV-images/per_frame/{scene_name}/clip_{clip_num}"
output_dir = f"/mnt/HDD1/iccv/orion_workspace/Bench2Drive/Orion/data/GIFs/clips/BEV/clip_{clip_num}.gif"
frame_duration_ms = 500  

image_paths = sorted(glob.glob(os.path.join(input_dir, "*.jpg")))

frames = []
print(f"Finding images in: {input_dir}")
for path in image_paths:
    frames.append(Image.open(path).convert("RGB"))

if frames:
    os.makedirs(os.path.dirname(output_dir), exist_ok=True)
    frames[0].save(
        output_dir,
        save_all=True,
        append_images=frames[1:],
        duration=frame_duration_ms,
        loop=0
    )
    print(f"GIF successfully saved to: {output_dir}")
else:
    print(f"No images (*.jpg) found in {input_dir}")
```
